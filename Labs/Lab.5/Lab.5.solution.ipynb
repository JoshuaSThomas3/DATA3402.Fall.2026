1. Create a `matrix` class with the following properties:
    * It can be initialized in 2 ways:
        1. with arguments `n` and `m`, the size of the matrix. A newly instanciated matrix will contain all zeros.
        2. with a list of lists of values. Note that since we are using lists of lists to implement matrices, it is possible that not all rows have the same number of columns. Test explicitly that the matrix is properly specified.
    * Matrix instances `M` can be indexed with `M[i][j]` and `M[i,j]`.
    * Matrix assignment works in 2 ways:
        1. If `M_1` and `M_2` are `matrix` instances `M_1=M_2` sets the values of `M_1` to those of `M_2`, if they are the same size. Error otherwise.
        2. In example above `M_2` can be a list of lists of correct size.







```python

# EX 1
# Creating a matrix class using nested lists

#I created a class matrix and then used nested python lists, the class can create a matrix filled with zeros or use values from existing lists
# I checked that all rows will have the same number of columns.
# I added indexing methods so i could acces an changge the matrix values if needed
# I also created a method assign to copy values from another matrix when the dimensions match.


class matrix:
    def __init__(self, n, m=None):

        # Create a matrix from a list of lists
        if isinstance(n, list):
            if len(n) == 0 or len(n[0]) == 0:
                raise ValueError("Matrix cannot be empty")

            columns = len(n[0])

            # Check that every row has the same length
            for row in n:
                if len(row) != columns:
                    raise ValueError("All rows must have the same length")

            # Copy the values into a new list
            self.data = []

            for row in n:
                self.data.append(row.copy())

        # Create a matrix filled with zeros
        else:
            if not isinstance(n, int) or not isinstance(m, int):
                raise ValueError("Dimensions must be integers")

            if n <= 0 or m <= 0:
                raise ValueError("Dimensions must be positive")

            self.data = []

            for i in range(n):
                row = []

                for j in range(m):
                    row.append(0)

                self.data.append(row)

    # Access elements using M[i][j] or M[i,j]
    
    # Access matrix elements and support slicing
    def __getitem__(self, index):

        # Handle M[i,j] and M[i:j,k:l]
        if isinstance(index, tuple):
            i, j = index

            # If either index is a slice
            if isinstance(i, slice) or isinstance(j, slice):

                rows = self.data[i] if isinstance(i, slice) else [self.data[i]]

                new_data = []

                for row in rows:
                    if isinstance(j, slice):
                        new_data.append(row[j])
                    else:
                        new_data.append([row[j]])

                return matrix(new_data)

            return self.data[i][j]

        # Handle M[i:j] by returning a matrix
        if isinstance(index, slice):
            return matrix(self.data[index])

        # Handle M[i][j] by returning a row
        return self.data[index]


    # Change an element using M[i,j] = value
    def __setitem__(self, index, value):
        if isinstance(index, tuple):
            i, j = index
            self.data[i][j] = value
        else:
            self.data[index] = value

    # Copy values from another matrix or list of lists
    def assign(self, other):
        if isinstance(other, matrix):
            values = other.data
        else:
            values = other

        # Validate the input using the matrix constructor
        new_matrix = matrix(values)

        # Check that dimensions match
        if len(new_matrix.data) != len(self.data):
            raise ValueError("Matrix dimensions do not match")

        if len(new_matrix.data[0]) != len(self.data[0]):
            raise ValueError("Matrix dimensions do not match")

        # Copy the values
        for i in range(len(self.data)):
            for j in range(len(self.data[0])):
                self.data[i][j] = new_matrix.data[i][j]


    
    # Return the number of rows and columns
    def shape(self):
        rows = len(self.data)
        columns = len(self.data[0])
        return (rows, columns)

    # Return the transpose of the matrix
    def transpose(self):
        rows, columns = self.shape()
        new_data = []

        for j in range(columns):
            new_row = []

            for i in range(rows):
                new_row.append(self.data[i][j])

            new_data.append(new_row)

        return matrix(new_data)

    # Return a single row as a matrix
    def row(self, n):
        return matrix([self.data[n].copy()])

    # Return a single column as a matrix
    def column(self, n):
        new_data = []

        for row in self.data:
            new_data.append([row[n]])

        return matrix(new_data)

    # Convert matrix to a list of lists
    def to_list(self):
        new_list = []

        for row in self.data:
            new_list.append(row.copy())

        return new_list

    # Extract part of the matrix
    # n0:n1 selects columns, m0:m1 selects rows
    def block(self, n0, n1, m0, m1):
        new_data = []

        for i in range(m0, m1):
            new_row = []

            for j in range(n0, n1):
                new_row.append(self.data[i][j])

            new_data.append(new_row)

        return matrix(new_data)


    def scalarmul(self, c):
        result = []

        for row in self.data:
            new_row = []

            for value in row:
                new_row.append(value * c)

            result.append(new_row)

        return matrix(result)

```

3. Write functions that create special matrices (note these are standalone functions, not member functions of your `matrix` class):
    * `constant(n,m,c)`: returns a `n` by `m` matrix filled with floats of value `c`.
    * `zeros(n,m)` and `ones(n,m)`: return `n` by `m` matrices filled with floats of value `0` and `1`, respectively.
    * `eye(n)`: returns the n by n identity matrix.

```python

# Ex  3


# Using function I generated different kinds of matrices.
# I created constant function in order to fill a matrix with a given float value.
# the function zeros and ones will use th constant function to a matrices which will only have the zeros or ones within them.
# Then using a nested loop, created an identity matrix, which will have ones on the main diagonal and zeros everywhere else.





# Created matrix filled with a constant value
def constant(n, m, c):
    values = []

    for i in range(n):
        row = []

        for j in range(m):
            row.append(float(c))

        values.append(row)

    return matrix(values)


# Create a matrix filled with zeros
def zeros(n, m):
    return constant(n, m, 0)


# Create a matrix filled with ones
def ones(n, m):
    return constant(n, m, 1)


# Create an identity matrix
def eye(n):
    values = []

    for i in range(n):
        row = []

        for j in range(n):
            if i == j:
                row.append(1.0)
            else:
                row.append(0.0)

        values.append(row)

    return matrix(values)
```

```python
#EX4

# I added alot of mathematical operations to my matrix class such as:
# The Method Scalar multiplications which would multiply every element with one number
# The Method Addition and subtraction but only if the matrix positions are matching positions.
# The Method  Matrix multiplication with mat_mult which uses loops to multiply rows by the colums and then add resulting products together.
# The Method Element wise multiplication so that matching elements can be multiplied.
# The Method equals will check if two matrices are equal.
# Dimension checks were include so that invalid operations wouldn't be ran.




#  All matrix classes combines

class matrix:

    def __init__(self, n, m=None):
        if isinstance(n, list):
            if len(n) == 0 or len(n[0]) == 0:
                raise ValueError("Matrix cannot be empty")

            for row in n:
                if len(row) != len(n[0]):
                    raise ValueError("Rows must have equal length")

            self.data = [row.copy() for row in n]

        else:
            if not isinstance(n, int) or not isinstance(m, int):
                raise ValueError("Dimensions must be integers")
            if n <= 0 or m <= 0:
                raise ValueError("Dimensions must be positive")

            self.data = [[0 for j in range(m)] for i in range(n)]

    # Access values and slices
    def __getitem__(self, index):
        if isinstance(index, tuple):
            i, j = index

            if isinstance(i, slice) or isinstance(j, slice):
                rows = self.data[i] if isinstance(i, slice) else [self.data[i]]
                result = []

                for row in rows:
                    if isinstance(j, slice):
                        result.append(row[j])
                    else:
                        result.append([row[j]])

                return matrix(result)

            return self.data[i][j]

        if isinstance(index, slice):
            return matrix(self.data[index])

        return self.data[index]

    # Change values
    def __setitem__(self, index, value):
        if isinstance(index, tuple):
            i, j = index
            self.data[i][j] = value
        else:
            self.data[index] = value

    # Assign values from another matrix
    def assign(self, other):
        if isinstance(other, matrix):
            values = other.to_list()
        else:
            values = matrix(other).to_list()

        if self.shape() != matrix(values).shape():
            raise ValueError("Matrix dimensions do not match")

        self.data = [row.copy() for row in values]

    # Exercise 2 - Matrix methods
    def shape(self):
        return (len(self.data), len(self.data[0]))

    def to_list(self):
        return [row.copy() for row in self.data]

    def transpose(self):
        rows, cols = self.shape()
        result = []

        for j in range(cols):
            new_row = []
            for i in range(rows):
                new_row.append(self[i, j])
            result.append(new_row)

        return matrix(result)

    def row(self, n):
        return matrix([self.data[n].copy()])

    def column(self, n):
        result = []

        for row in self.data:
            result.append([row[n]])

        return matrix(result)

    def block(self, n0, n1, m0, m1):
        result = []

        for i in range(m0, m1):
            result.append(self.data[i][n0:n1])

        return matrix(result)

    # EX 4 Scalar multiplication
    def scalarmul(self, c):
        result = []

        for row in self.data:
            new_row = []

            for value in row:
                new_row.append(value * c)

            result.append(new_row)

        return matrix(result)

    # Matrix addition
    def add(self, N):
        if self.shape() != N.shape():
            raise ValueError("Matrix dimensions do not match")

        result = []

        for i in range(self.shape()[0]):
            row = []
            for j in range(self.shape()[1]):
                row.append(self[i, j] + N[i, j])
            result.append(row)

        return matrix(result)

    # Matrix subtraction
    def sub(self, N):
        if self.shape() != N.shape():
            raise ValueError("Matrix dimensions do not match")

        result = []

        for i in range(self.shape()[0]):
            row = []
            for j in range(self.shape()[1]):
                row.append(self[i, j] - N[i, j])
            result.append(row)

        return matrix(result)

    # matrix multiplication
    def mat_mult(self, N):
        rows_A, cols_A = self.shape()
        rows_B, cols_B = N.shape()

        if cols_A != rows_B:
            raise ValueError("Incompatible matrix dimensions")

        result = []

        for i in range(rows_A):
            row = []

            for j in range(cols_B):
                total = 0

                for k in range(cols_A):
                    total += self[i, k] * N[k, j]

                row.append(total)

            result.append(row)

        return matrix(result)

    # Element-wise multiplication
    def element_mult(self, N):
        if self.shape() != N.shape():
            raise ValueError("Matrix dimensions do not match")

        result = []

        for i in range(self.shape()[0]):
            row = []

            for j in range(self.shape()[1]):
                row.append(self[i, j] * N[i, j])

            result.append(row)

        return matrix(result)

    # Compare two matrices
    def equals(self, N):
        if not isinstance(N, matrix):
            return False

        if self.shape() != N.shape():
            return False

        for i in range(self.shape()[0]):
            for j in range(self.shape()[1]):
                if self[i, j] != N[i, j]:
                    return False

        return True



    
    # Exercise 5 

    def __add__(self, other):
        return self.add(other)

    def __sub__(self, other):
        return self.sub(other)

    def __mul__(self, other):
        if isinstance(other, matrix):
            return self.mat_mult(other)

        elif isinstance(other, (int, float)):
            return self.scalarmul(other)

        return NotImplemented

    def __rmul__(self, other):
        if isinstance(other, (int, float)):
            return self.scalarmul(other)

        return NotImplemented

    def __eq__(self, other):
        return self.equals(other)

```

```python

# Exercise 6
# Tested the four basic properties of matrix multiplication. 

# by changing the grouping I was able to check associativity.
# Then tested distributivity by using adding and multiplication.
# Compared B * A with A * B to show the matrix

# then laslty I multiplied A by an identity matrix in order to check and see if the original matrix changed or not
# using assertions I confirmed results.



# Create two 2 by 2 matrices
A = matrix([[1, 2],
            [3, 4]])

B = matrix([[2, 0],
            [1, 3]])

# Create a third matrix using A and B
C = A + B

# Create a 2 by 2 identity matrix
I = eye(2)

print("Matrix A:", A.to_list())
print("Matrix B:", B.to_list())
print("Matrix C:", C.to_list())


# Property 1: Associativity
left1 = (A * B) * C
right1 = A * (B * C)

print("\n1. Associativity")
print("(A * B) * C =", left1.to_list())
print("A * (B * C) =", right1.to_list())
print("Equal:", left1 == right1)


# Property 2: Distributivity
left2 = A * (B + C)
right2 = (A * B) + (A * C)

print("\n2. Distributivity")
print("A * (B + C) =", left2.to_list())
print("A * B + A * C =", right2.to_list())
print("Equal:", left2 == right2)


# Property 3: Matrix multiplication is not commutative
AB = A * B
BA = B * A

print("\n3. Noncommutativity")
print("A * B =", AB.to_list())
print("B * A =", BA.to_list())
print("Different:", AB != BA)


# Property 4: Identity matrix
result = A * I

print("\n4. Identity")
print("A * I =", result.to_list())
print("Equal to A:", result == A)
```
