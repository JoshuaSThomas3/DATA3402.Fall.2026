*Exercise 4:* Write a function that takes an `n×n` matrix representing a tic‑tac‑toe game and returns one of the following values:

- `-1` if the game is **incomplete** (still empty spaces and no winner)
- `0` if the game is a **draw**
- `1` if **player 1** (`"X"`) has won
- `2` if **player 2** (`"O"`) has won

Here are some example inputs you can use to test your code:


````python
# Write your solution here

# Exercise 4: Check the status of a Tic-Tac-Toe game.

## Created a functino that checks the current status of a tic-tac-toe game using the values un the board.
# Used loops so i can get all the rows and columns and diagonals to check for a winner.
# all() function was used to determine win or loss based on 1's and 2's in a row. No winner is when there is no value just empty.
# function will also return a 0 to display a draw
# I also added some tests in order to check if the identification of the game status is win, draw, loss, using assert statements.

def check_winner(board):
    # Find the size of the board.
    # For a 3x3 board, n = 3.
    n = len(board)

    
    lines = []

    
    for row in board:
        lines.append(row)

    
    for col in range(n):
        # Collect values from the same column.
        column = [board[row][col] for row in range(n)]

        # Add this column to the list of winning lines.
        lines.append(column)

    
    main_diagonal = [board[i][i] for i in range(n)]
    lines.append(main_diagonal)

    
    other_diagonal = [board[i][n - 1 - i] for i in range(n)]
    lines.append(other_diagonal)

    
    for line in lines:

        
        if all(value == 1 for value in line):
            return 1

        
        if all(value == 2 for value in line):
            return 2

    
    for row in board:
        if 0 in row:
            return -1

    
    return 0
````

*Exercise 5:* Write a function that takes a game board, a player number, and `(row, col)` coordinates and places the correct mark (`"X"` or `"O"`) in that location.

Requirements:

- Only allow placing a mark in a previously empty location.
- Return `True` if the move was successful, and `False` otherwise.


````python
# Write your solution here

# created a function that the player can use to place their mark on the board using row column coordinates
# Using if statements, I checked that the player number number and coordinates are valid before making a move
# It will also check if there is already a value there too, in which case it won't change the board and return false.
# Otherwise it will input the value and return with displaying the board updates.


def place_mark(board, player, row, col):

    n = len(board)

    if player not in (1, 2):
        return False

    if row < 0 or row >= n:
        return False

    if col < 0 or col >= len(board[row]):
        return False

    if board[row][col] != 0:
        return False

    board[row][col] = player

    return True

    
````

*Exercise 7:* Write a function that takes a board, a player number, and a location string (as in Exercise 6), then uses your function from Exercise 5 to update the board.

I made a function that will convert B2 into numerical coordinates, it also seperates row letter from the column number and checks it the format is correct. The it will use the place_mark function to perform the move.

````python
# Write your solution here

# Created function that allows player to input a location like "B2".
# strip() and upper() were used so it will accept lower and upper case.
# this function can check if the format is correct by seperating the row letter from the column number.
# ord() was used to convert from letter to row index and int() to converth the column into a number
# place_mark() will check move and update the board.

def place_location(board, player, location):
    
    location = location.strip().upper()

   
    if len(location) < 2:
        return False

    
    row_letter = location[0]
    column_text = location[1:]

    
    if not row_letter.isalpha():
        return False

   
    if not column_text.isdigit():
        return False

    row = ord(row_letter) - ord("A")

   
    col = int(column_text) - 1

  
    return place_mark(board, player, row, col)
````

*Exercise 11:* Develop a version of the game where one player is the computer.

Note: you do **not** need an extensive search for the best move. For example, you can have the computer:
- block obvious losses
- otherwise try to create a winning row/column/diagonal


````python
# Write your solution here

# Created a basic coputer player that makes decision based on the current game board.
# available_mmoves() finds all the empty spaces where a move is posssible

# computer_basic() loops were used to check possible move in order for the computer to win.
# if there isn't a winning move then it will block opponents moves. Otherwise it will just choose a any space.

# testing was done using two boards one to choose the best move to win, and another was to block a winning move from the opponent.


import random


def available_moves(board):
    
    moves = []

    
    for row in range(len(board)):

        
        for col in range(len(board[row])):

            
            if board[row][col] == 0:
                moves.append((row, col))

    
    return moves


def computer_basic(board, player):
    
    opponent = 3 - player

    
    moves = available_moves(board)

    
    for row, col in moves:

        trial = [r.copy() for r in board]

        place_mark(trial, player, row, col)

        if check_winner(trial) == player:
            return (row, col)

    for row, col in moves:

        trial = [r.copy() for r in board]

       
        place_mark(trial, opponent, row, col)

        
        if check_winner(trial) == opponent:
            return (row, col)

    
    if moves:
        return random.choice(moves)

    return None
````

*Exercise 12:* Develop a version of the game where one player is the computer. This time, write a computer player using exhaustive search with a max depth parameter, similar to lecture.

I created a smart computer player using minimax algorithm, this one considers all the outcomes instead of the fastest method of victory.
Using a recursive function was my next step in order to create a maximization of points for the computer and minimilization for the opponent. Max depth parameter was used to limit the search quantity that way it wouldn't keep looking for possibilities for too long that a decision wouldn't be made. Evaluation function gives positive for favorable scores and negative for unfavorable scores.

````python
# Write your solution here


# minimax was used to create a smarter computer player to look ahead for future moves. And evaluate_board() to give score to determin if winning or losing.
# minimax function uses a recursion to test both players moves and maximize computers winning chance and minimize players chance.
# max_depth to limit how many moves ahead the computer searches that way it doesnt get stuck searching.
# compute_smart function compares scores and moves based on the highest score. Same tesing using assert statements


def evaluate_board(board, player):
    # Check if either player has won
    winner = check_winner(board)

    if winner == player:
        return 100

    if winner == 3 - player:
        return -100

    if winner == 0:
        return 0

   
    score = 0
    n = len(board)

    # Make a list of all rows, columns, and diagonals
    lines = []

    for row in board:
        lines.append(row)

    for col in range(n):
        column = []
        for row in range(n):
            column.append(board[row][col])
        lines.append(column)

    diagonal1 = []
    diagonal2 = []

    for i in range(n):
        diagonal1.append(board[i][i])
        diagonal2.append(board[i][n - 1 - i])

    lines.append(diagonal1)
    lines.append(diagonal2)

    
    for line in lines:
        my_marks = line.count(player)
        other_marks = line.count(3 - player)

        if other_marks == 0:
            score += my_marks

        if my_marks == 0:
            score -= other_marks

    return score


def minimax(board, player, computer, depth, max_depth):
    
    winner = check_winner(board)

    if winner != -1 or depth >= max_depth:
        return evaluate_board(board, computer)

    
    if player == computer:
        best_score = float("-inf")
    else:
        best_score = float("inf")

    
    for row, col in available_moves(board):
        
        test_board = [r.copy() for r in board]
        test_board[row][col] = player

        
        score = minimax(
            test_board,
            3 - player,
            computer,
            depth + 1,
            max_depth
        )

        if player == computer:
            best_score = max(best_score, score)
        else:
            best_score = min(best_score, score)

    return best_score


def computer_smart(board, player, max_depth=2):
    
    moves = available_moves(board)

    if len(moves) == 0:
        return None

    best_move = moves[0]
    best_score = float("-inf")

    
    for row, col in moves:
        test_board = [r.copy() for r in board]
        test_board[row][col] = player

        
        score = minimax(
            test_board,
            3 - player,
            player,
            1,
            max_depth
        )

        if score > best_score:
            best_score = score
            best_move = (row, col)

    return best_move


def play_smart_computer(n=3, max_depth=2):
    
    board = create_board(n)
    player = 1

    while check_winner(board) == -1:
        display_labeled_board(board)

        if player == 1:
            player_turn(board, player)
        else:
            move = computer_smart(board, player, max_depth)
            row, col = move
            place_mark(board, player, row, col)
            print("Computer moved to:", move)

        
        player = 3 - player

    # Display the final result
    display_labeled_board(board)

    winner = check_winner(board)

    if winner == 1:
        print("Human wins!")
    elif winner == 2:
        print("Computer wins!")
    else:
        print("The game is a draw!")
````

*Exercise 13:* Make the 2 computer players play each-other for 10 games on a 3x3, then 4x4, then 5x5 grid. Set the max depth so that the games only take seconds. Measure the "smarter" player's win rate for each grid.

I ran 10 games on each board size using basic computer player vs. smarter minimax computer.

3x3: 60%
4x4: 20%

````python
# Write your solution here


# created a function that puts the basic coomputer against the smart minimax computer and they play each other.
# smart computer will uses minimax to choose moves, basic computer checks for immediate wins or blocks or random moves.
# while loop will continue the game until someone wins or draws.
# and then a function will return result of the game.



def computer_game(n, max_depth=2):
    
    board = create_board(n)

    
    player = 1

    
    while check_winner(board) == -1:

        if player == 1:
            move = computer_smart(board, player, max_depth)
        else:
            move = computer_basic(board, player)

        
        if move is None:
            break

        
        row, col = move
        place_mark(board, player, row, col)

        
        player = 3 - player

    return check_winner(board)

    
# Created another function which tests both computer players over multiple games. Using for loop I ran 10 games for each board size.
# Calculate the win rate of smart computer by dividing wins by total number of games and multiplying by 100 to get a percentage
# results show smart computer wuns 60% of 3x3 board games, 20% on 4x4 and 0% on a 5x5 board. 
# since we limit the depth of how long the smart computer can consider all moves it will lose in bigger boards. And it was only 10 games.

def test_computers(n, games=10, max_depth=2):
    
    smart_wins = 0
    basic_wins = 0
    draws = 0

    
    for i in range(games):
        winner = computer_game(n, max_depth)

        if winner == 1:
            smart_wins += 1
        elif winner == 2:
            basic_wins += 1
        else:
            draws += 1

    
    win_rate = smart_wins / games * 100

    print("Board size:", n, "x", n)
    print("Smart computer wins:", smart_wins)
    print("Basic computer wins:", basic_wins)
    print("Draws:", draws)
    print("Smart computer win rate:", win_rate, "%")
    print()

    return win_rate
````
