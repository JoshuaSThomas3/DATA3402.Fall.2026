1. Write a "counter" class that can be incremented up to a specified maximum value, will print an error if an attempt is made to increment beyond that value, and allows reseting the counter. 

```python
# Created a counter class that begins with zero and used uncrement method to increase value by one
# and it will stay below maximum. If the counter reaches the limit, it prints an error.
# reset method will return the counter to zero.




class Counter:
    def __init__(self, maximum):
        # Start the counter at zero
        self.value = 0
        self.maximum = maximum

    def increment(self):
        # Increase the counter if it is below the maximum
        if self.value < self.maximum:
            self.value += 1
        else:
            print("Error: Counter has reached the maximum value.")

    def reset(self):
        # Set the counter back to zero
        self.value = 0
```

```python

# Exercise 1 - Test 

# Create a counter with a maximum value of 3
counter = Counter(3)

# Increase the counter three times
counter.increment()
counter.increment()
counter.increment()

print("Counter value:", counter.value)

# Try to increase beyond the maximum
counter.increment()


counter.reset()

print("Value after reset:", counter.value)
```

2. Copy and paste your solution to question 1 and modify it so that all the data held by the counter is private. Implement functions to check the value of the counter, check the maximum value, and check if the counter is at the maximum.

```python

# Exercise 2 

# Modified COunter class to use privae variables for current and max value.
#Used double underscores to prevent direct access and added methods to retrieve these values and check.
# whether the counter has reached its maximum

class Counter:
    def __init__(self, maximum):
        # Store the counter value and maximum privately
        self.__value = 0
        self.__maximum = maximum

    def increment(self):
        # Only increase the value if it is below the maximum
        if self.__value < self.__maximum:
            self.__value += 1
        else:
            print("Error: Counter has reached the maximum.")

    def reset(self):
        # Reset the counter back to zero
        self.__value = 0

    def get_value(self):
        # Return the current counter value
        return self.__value

    def get_maximum(self):
        # Return the maximum allowed value
        return self.__maximum

    def at_maximum(self):
        # Return True if the counter reached the maximum
        return self.__value == self.__maximum
```

```python

# Exercise 2 - Test the Counter class

# Create a counter with a maximum of 3
counter = Counter(3)

print("Starting value:", counter.get_value())
print("Maximum value:", counter.get_maximum())
print("At maximum?", counter.at_maximum())

# Increase the counter three times
counter.increment()
counter.increment()
counter.increment()

print("Value after increments:", counter.get_value())
print("At maximum?", counter.at_maximum())

# Try to go past the maximum
counter.increment()

# Reset the counter to zero
counter.reset()

print("Value after reset:", counter.get_value())
print("At maximum?", counter.at_maximum())
```

3. Implement a class to represent a rectangle, holding the length, width, and $x$ and $y$ coordinates of a corner of the object. Implement functions that compute the area and perimeter of the rectangle. Make all data members private and provide accessors to retrieve values of data members. 

```python

# Exercise 3 - Rectangle class

# Created Rectangle class which stores length, width, and coordinates of a corner as private values
#Used length times width to calculate area, and 2 times the sum of both sides for perimeter.
# used get methods to retrive said stored values




class Rectangle:
    def __init__(self, length, width, x, y):
        # Store the rectangle information privately
        self.__length = length
        self.__width = width
        self.__x = x
        self.__y = y

    # Calculate the area of the rectangle
    def area(self):
        return self.__length * self.__width

    # Calculate the perimeter of the rectangle
    def perimeter(self):
        return 2 * (self.__length + self.__width)

    # Get the length
    def get_length(self):
        return self.__length

    # Get the width
    def get_width(self):
        return self.__width

    # Get the x coordinate
    def get_x(self):
        return self.__x

    # Get the y coordinate
    def get_y(self):
        return self.__y
```

```python

# Exercise 5 - 

# Created a shape base class with methods shared by different geometric shapes.
# Used raise method and NotImplementedError until there is defined value in the child classes
#  Made Rectangle and Circle inherit from Shape class then give their own area and perimeter calculation.





# Base class for different shapes
class Shape:
    def area(self):
        raise NotImplementedError("Area not implemented")

    def perimeter(self):
        raise NotImplementedError("Perimeter not implemented")

    def get_x(self):
        raise NotImplementedError("X coordinate not implemented")

    def get_y(self):
        raise NotImplementedError("Y coordinate not implemented")


# Rectangle inherits from Shape
class Rectangle(Shape):
    def __init__(self, length, width, x, y):
        self.__length = length
        self.__width = width
        self.__x = x
        self.__y = y

    # Calculate the rectangle's area
    def area(self):
        return self.__length * self.__width

    # Calculate the rectangle's perimeter
    def perimeter(self):
        return 2 * (self.__length + self.__width)

    def get_length(self):
        return self.__length

    def get_width(self):
        return self.__width

    def get_x(self):
        return self.__x

    def get_y(self):
        return self.__y


# Circle also inherits from Shape
class Circle(Shape):
    def __init__(self, radius, x, y):
        self.__radius = radius
        self.__x = x
        self.__y = y

    # Calculate the circle's area
    def area(self):
        return math.pi * self.__radius ** 2

    # Calculate the circumference
    def perimeter(self):
        return 2 * math.pi * self.__radius

    def get_radius(self):
        return self.__radius

    def get_x(self):
        return self.__x

    def get_y(self):
        return self.__y
```

6. Implement a triangle class analogous to the rectangle and circle in question 5.

```python

# Exercise 6 

## Triangle inherits from shape class, stored the coordinate values from all 3 corners.
## Area was calculated using the coordinates and perimeter was distance between the corners.
## Using method we can retrieve the coordinates with get_points.


import math

class Triangle(Shape):
    def __init__(self, x1, y1, x2, y2, x3, y3):
        # Store the three corners of the triangle
        self.__x1 = x1
        self.__y1 = y1
        self.__x2 = x2
        self.__y2 = y2
        self.__x3 = x3
        self.__y3 = y3

    # Calculate the area using the coordinates
    def area(self):
        area = abs(
            self.__x1 * (self.__y2 - self.__y3)
            + self.__x2 * (self.__y3 - self.__y1)
            + self.__x3 * (self.__y1 - self.__y2)
        ) / 2

        return area

    # Calculate the perimeter using the three sides
    def perimeter(self):
        side1 = math.sqrt((self.__x2 - self.__x1)**2
                          + (self.__y2 - self.__y1)**2)

        side2 = math.sqrt((self.__x3 - self.__x2)**2
                          + (self.__y3 - self.__y2)**2)

        side3 = math.sqrt((self.__x1 - self.__x3)**2
                          + (self.__y1 - self.__y3)**2)

        return side1 + side2 + side3

    # Return the first corner's coordinates
    def get_x(self):
        return self.__x1

    def get_y(self):
        return self.__y1

    # Return all three corners
    def get_points(self):
        return [
            (self.__x1, self.__y1),
            (self.__x2, self.__y2),
            (self.__x3, self.__y3)
        ]
```

```python

# Exercise 6 - Test the Triangle class

# Create a triangle with three corners
triangle = Triangle(0, 0, 3, 0, 0, 4)

# Display the coordinates
print("Triangle corners:", triangle.get_points())

# Calculate the area and perimeter
print("Area:", triangle.area())
print("Perimeter:", triangle.perimeter())

# Check that Triangle inherits from Shape
print("Is a Shape:", isinstance(triangle, Shape))

# Check the expected results
assert triangle.area() == 6.0
assert triangle.perimeter() == 12.0
assert isinstance(triangle, Shape)

print("Exercise 6 tests passed!")
```

7. Add a function to the object classes, including the base, that returns a list of up to 16 pairs of  $x$ and $y$ points on the perimeter of the object. 

```python

# Exercise 7 

## Added perimeter_points methods to the shape classes.
## the rectangle, circle, and triangle all generate 16 points along its four sides.
## Each method will return coordinate pairs in a list.


import math

class Shape:
    def area(self):
        raise NotImplementedError

    def perimeter(self):
        raise NotImplementedError

    
    # Check if two shapes overlap
    def overlaps(self, other):

        # Check points from this shape
        for x, y in self.perimeter_points():
            if other.is_inside(x, y):
                return True

        # Check points from the other shape
        for x, y in other.perimeter_points():
            if self.is_inside(x, y):
                return True

        return False


    def get_x(self):
        raise NotImplementedError

    def get_y(self):
        raise NotImplementedError

    def perimeter_points(self):
        raise NotImplementedError

    def is_inside(self, x, y):
        raise NotImplementedError


class Rectangle(Shape):
    def __init__(self, length, width, x, y):
        self.__length = length
        self.__width = width
        self.__x = x
        self.__y = y

    def area(self):
        return self.__length * self.__width

    def perimeter(self):
        return 2 * (self.__length + self.__width)

    def get_length(self):
        return self.__length

    def get_width(self):
        return self.__width

    def get_x(self):
        return self.__x

    def get_y(self):
        return self.__y

    def perimeter_points(self):
        points = []

        # Find 4 points along each side
        for i in range(4):
            t = i / 4

            points.append((self.__x + t * self.__length,
                           self.__y))

            points.append((self.__x + self.__length,
                           self.__y + t * self.__width))

            points.append((self.__x + self.__length * (1 - t),
                           self.__y + self.__width))

            points.append((self.__x,
                           self.__y + self.__width * (1 - t)))

        return points

    def is_inside(self, x, y):
        if (self.__x <= x <= self.__x + self.__length and
            self.__y <= y <= self.__y + self.__width):
            return True

        return False


class Circle(Shape):
    def __init__(self, radius, x, y):
        self.__radius = radius
        self.__x = x
        self.__y = y

    def area(self):
        return math.pi * self.__radius ** 2

    def perimeter(self):
        return 2 * math.pi * self.__radius

    def get_radius(self):
        return self.__radius

    def get_x(self):
        return self.__x

    def get_y(self):
        return self.__y

    def perimeter_points(self):
        points = []

        # Divide the circle into 16 angles
        for i in range(16):
            angle = 2 * math.pi * i / 16

            x = self.__x + self.__radius * math.cos(angle)
            y = self.__y + self.__radius * math.sin(angle)

            points.append((x, y))

        return points

    def is_inside(self, x, y):
        # Find the squared distance from the center
        distance = (x - self.__x)**2 + (y - self.__y)**2

        # Compare the distance with the radius squared
        if distance <= self.__radius**2:
            return True

        return False



class Triangle(Shape):
    def __init__(self, x1, y1, x2, y2, x3, y3):
        self.__x1 = x1
        self.__y1 = y1
        self.__x2 = x2
        self.__y2 = y2
        self.__x3 = x3
        self.__y3 = y3

    def area(self):
        result = abs(
            self.__x1 * (self.__y2 - self.__y3)
            + self.__x2 * (self.__y3 - self.__y1)
            + self.__x3 * (self.__y1 - self.__y2)
        ) / 2

        return result

    def perimeter(self):
        side1 = math.dist((self.__x1, self.__y1),
                          (self.__x2, self.__y2))

        side2 = math.dist((self.__x2, self.__y2),
                          (self.__x3, self.__y3))

        side3 = math.dist((self.__x3, self.__y3),
                          (self.__x1, self.__y1))

        return side1 + side2 + side3

    def get_x(self):
        return self.__x1

    def get_y(self):
        return self.__y1

    def get_points(self):
        return [(self.__x1, self.__y1),
                (self.__x2, self.__y2),
                (self.__x3, self.__y3)]

    
    def perimeter_points(self):
        points = []

        # Get the three corners of the triangle
        corners = self.get_points()

        # Go through all three sides
        for side in range(3):
            start = corners[side]
            end = corners[(side + 1) % 3]

            # Generate five points along each side
            for i in range(5):
                t = i / 5

                x = start[0] + t * (end[0] - start[0])
                y = start[1] + t * (end[1] - start[1])

                points.append((x, y))

        # Return all 15 coordinate pairs
        return points


    def is_inside(self, x, y):

        # Calculate the signs for the three sides
        d1 = (x - self.__x2) * (self.__y1 - self.__y2) - (self.__x1 - self.__x2) * (y - self.__y2)

        d2 = (x - self.__x3) * (self.__y2 - self.__y3) - (self.__x2 - self.__x3) * (y - self.__y3)

        d3 = (x - self.__x1) * (self.__y3 - self.__y1) - (self.__x3 - self.__x1) * (y - self.__y1)

        # Check whether the signs are mixed
        has_negative = d1 < 0 or d2 < 0 or d3 < 0
        has_positive = d1 > 0 or d2 > 0 or d3 > 0

        # Points inside should not have mixed signs
        return not (has_negative and has_positive)
```

```python

# Exercise 7 - Test perimeter points

# Create one of each shape
rect = Rectangle(8, 5, 2, 3)
circle = Circle(5, 2, 3)
triangle = Triangle(0, 0, 3, 0, 0, 4)

# Get the perimeter points
rect_points = rect.perimeter_points()
circle_points = circle.perimeter_points()
triangle_points = triangle.perimeter_points()

# Print the results
print("Rectangle points:", rect_points)
print("Number of rectangle points:", len(rect_points))

print("\nCircle points:", circle_points)
print("Number of circle points:", len(circle_points))

print("\nTriangle points:", triangle_points)
print("Number of triangle points:", len(triangle_points))

# Check that each shape returns at most 16 points
assert len(rect_points) == 16
assert len(circle_points) == 16
assert len(triangle_points) == 15

# Make sure every point contains x and y coordinates
for shape in [rect, circle, triangle]:
    for point in shape.perimeter_points():
        assert len(point) == 2

print("\nExercise 7 tests passed!")
```

```python
%%writefile paint.py



## Created a paint module using Canvas class which will store characters as rows and columns.
## Added paint methods for Rectangle, Circle, and triangle.

## each shape will draw its outline using different characters/symbols
## Created CompoundSHape to group shapes together and draw them all in one canvas.





import math


# Canvas class based on Lecture 9
class Canvas:
    def __init__(self, width, height):
        self.width = width
        self.height = height
        self.data = [[' '] * width for i in range(height)]

    def set_pixel(self, row, col, char='*'):
        # Only draw points that fit on the canvas
        row = int(round(row))
        col = int(round(col))

        if 0 <= row < self.height and 0 <= col < self.width:
            self.data[row][col] = char

    def get_pixel(self, row, col):
        return self.data[row][col]

    def clear_canvas(self):
        self.data = [[' '] * self.width for i in range(self.height)]

    def v_line(self, x, y, h, char='*'):
        for i in range(h):
            self.set_pixel(x + i, y, char)

    def h_line(self, x, y, w, char='*'):
        for i in range(w):
            self.set_pixel(x, y + i, char)

    def line(self, x1, y1, x2, y2, char='*'):
        # Draw points between the two endpoints
        steps = int(max(abs(x2 - x1), abs(y2 - y1)))

        if steps == 0:
            self.set_pixel(x1, y1, char)
            return

        for i in range(steps + 1):
            t = i / steps
            x = x1 + t * (x2 - x1)
            y = y1 + t * (y2 - y1)
            self.set_pixel(x, y, char)

    def display(self):
        for row in self.data:
            print(''.join(row))


# Base class for the different shapes
class Shape:
    def area(self):
        raise NotImplementedError

    def perimeter(self):
        raise NotImplementedError

    def perimeter_points(self):
        raise NotImplementedError

    def is_inside(self, x, y):
        raise NotImplementedError

    def overlaps(self, other):
        # Check sampled perimeter points from both shapes
        for x, y in self.perimeter_points():
            if other.is_inside(x, y):
                return True

        for x, y in other.perimeter_points():
            if self.is_inside(x, y):
                return True

        return False

    def paint(self, canvas):
        raise NotImplementedError


class Rectangle(Shape):
    def __init__(self, length, width, x, y):
        self.__length = length
        self.__width = width
        self.__x = x
        self.__y = y

    def area(self):
        return self.__length * self.__width

    def perimeter(self):
        return 2 * (self.__length + self.__width)

    def get_length(self):
        return self.__length

    def get_width(self):
        return self.__width

    def get_x(self):
        return self.__x

    def get_y(self):
        return self.__y

    def perimeter_points(self):
        points = []

        for i in range(4):
            t = i / 4
            points.append((self.__x + t * self.__length,
                           self.__y))
            points.append((self.__x + self.__length,
                           self.__y + t * self.__width))
            points.append((self.__x + (1 - t) * self.__length,
                           self.__y + self.__width))
            points.append((self.__x,
                           self.__y + (1 - t) * self.__width))

        return points

    def is_inside(self, x, y):
        return (self.__x <= x <= self.__x + self.__length
                and self.__y <= y <= self.__y + self.__width)

    def paint(self, canvas):
        # Draw the four sides of the rectangle
        x = self.__x
        y = self.__y
        length = self.__length
        width = self.__width

        canvas.line(y, x, y, x + length, '#')
        canvas.line(y + width, x, y + width, x + length, '#')
        canvas.line(y, x, y + width, x, '#')
        canvas.line(y, x + length, y + width, x + length, '#')


class Circle(Shape):
    def __init__(self, radius, x, y):
        self.__radius = radius
        self.__x = x
        self.__y = y

    def area(self):
        return math.pi * self.__radius ** 2

    def perimeter(self):
        return 2 * math.pi * self.__radius

    def get_radius(self):
        return self.__radius

    def get_x(self):
        return self.__x

    def get_y(self):
        return self.__y

    def perimeter_points(self):
        points = []

        for i in range(16):
            angle = 2 * math.pi * i / 16
            x = self.__x + self.__radius * math.cos(angle)
            y = self.__y + self.__radius * math.sin(angle)
            points.append((x, y))

        return points

    def is_inside(self, x, y):
        distance = (x - self.__x)**2 + (y - self.__y)**2
        return distance <= self.__radius**2

    def paint(self, canvas):
        # Use small line segments around the circle
        for i in range(64):
            a1 = 2 * math.pi * i / 64
            a2 = 2 * math.pi * (i + 1) / 64

            x1 = self.__x + self.__radius * math.cos(a1)
            y1 = self.__y + self.__radius * math.sin(a1)
            x2 = self.__x + self.__radius * math.cos(a2)
            y2 = self.__y + self.__radius * math.sin(a2)

            canvas.line(y1, x1, y2, x2, 'O')


class Triangle(Shape):
    def __init__(self, x1, y1, x2, y2, x3, y3):
        self.__points = [(x1, y1), (x2, y2), (x3, y3)]

    def get_points(self):
        return self.__points.copy()

    def get_x(self):
        return self.__points[0][0]

    def get_y(self):
        return self.__points[0][1]

    def area(self):
        (x1, y1), (x2, y2), (x3, y3) = self.__points
        return abs(x1*(y2-y3) + x2*(y3-y1)
                   + x3*(y1-y2)) / 2

    def perimeter(self):
        total = 0

        for i in range(3):
            p1 = self.__points[i]
            p2 = self.__points[(i + 1) % 3]
            total += math.dist(p1, p2)

        return total

    def perimeter_points(self):
        points = []

        for side in range(3):
            start = self.__points[side]
            end = self.__points[(side + 1) % 3]

            for i in range(5):
                t = i / 5
                x = start[0] + t * (end[0] - start[0])
                y = start[1] + t * (end[1] - start[1])
                points.append((x, y))

        return points

    def is_inside(self, x, y):
        # Compare areas of the three smaller triangles
        p1, p2, p3 = self.__points

        def sign(a, b, c):
            return ((a[0] - c[0]) * (b[1] - c[1])
                    - (b[0] - c[0]) * (a[1] - c[1]))

        point = (x, y)
        d1 = sign(point, p1, p2)
        d2 = sign(point, p2, p3)
        d3 = sign(point, p3, p1)

        negative = d1 < 0 or d2 < 0 or d3 < 0
        positive = d1 > 0 or d2 > 0 or d3 > 0

        return not (negative and positive)

    def paint(self, canvas):
        # Connect the three corners with lines
        for i in range(3):
            x1, y1 = self.__points[i]
            x2, y2 = self.__points[(i + 1) % 3]
            canvas.line(y1, x1, y2, x2, '+')


class CompoundShape(Shape):
    def __init__(self, shapes):
        self.shapes = shapes

    def paint(self, canvas):
        # Paint every shape in the group
        for shape in self.shapes:
            shape.paint(canvas)
```

11. Create a `RasterDrawing` class. Demonstrate that you can create a drawing made of several shapes, paint the drawing, modify the drawing, and paint it again. 

```python

%%writefile -a paint.py
import json

## Created Raster drawing class to store multiple shapes
## Using a dictionary I organized the shapes by their names
## Created add_shape and remove_shape methods to add and remove shapes from the drawing.
# Paint method will draw each stored shape on the canvas while the update method wil clear the canvas before drawing.
# In my test I removed the circle and displayed the result to see if it works.





# Exercise 11 - RasterDrawing class

class RasterDrawing:
    def __init__(self):
        # Store the shapes by their names
        self.shapes = {}
        self.shape_names = []

    def add_shape(self, shape, name=None):
        # Give the shape a name if one was not provided
        if name is None:
            name = "shape_" + str(len(self.shapes))

        # Add the shape to the dictionary
        self.shapes[name] = shape

        if name not in self.shape_names:
            self.shape_names.append(name)

    def remove_shape(self, name):
        # Remove a shape if it exists
        if name in self.shapes:
            del self.shapes[name]
            self.shape_names.remove(name)
        else:
            print("Shape not found.")

    def paint(self, canvas):
        # Draw every shape stored in the drawing
        for name in self.shape_names:
            shape = self.shapes[name]
            shape.paint(canvas)

    def update(self, canvas):
        # Clear the canvas and draw the shapes again
        canvas.clear_canvas()
        self.paint(canvas)

    @staticmethod
    def _shape_to_dict(shape):
        if isinstance(shape, Rectangle):
            return {
                "type": "Rectangle",
                "length": shape.get_length(),
                "width": shape.get_width(),
                "x": shape.get_x(),
                "y": shape.get_y()
            }

        if isinstance(shape, Circle):
            return {
                "type": "Circle",
                "radius": shape.get_radius(),
                "x": shape.get_x(),
                "y": shape.get_y()
            }

        if isinstance(shape, Triangle):
            return {
                "type": "Triangle",
                "points": shape.get_points()
            }

        if isinstance(shape, CompoundShape):
            return {
                "type": "CompoundShape",
                "shapes": [
                    RasterDrawing._shape_to_dict(s)
                    for s in shape.shapes
                ]
            }

        raise TypeError("Unsupported shape type")

    # Recreate a shape from its saved information
    @staticmethod
    def _dict_to_shape(data):
        kind = data["type"]

        if kind == "Rectangle":
            return Rectangle(
                data["length"], data["width"],
                data["x"], data["y"]
            )

        if kind == "Circle":
            return Circle(
                data["radius"], data["x"], data["y"]
            )

        if kind == "Triangle":
            p = data["points"]
            return Triangle(
                p[0][0], p[0][1],
                p[1][0], p[1][1],
                p[2][0], p[2][1]
            )

        if kind == "CompoundShape":
            shapes = [
                RasterDrawing._dict_to_shape(s)
                for s in data["shapes"]
            ]
            return CompoundShape(shapes)

        raise ValueError("Unknown shape type: " + str(kind))

    # Save the names and information for every shape
    def save(self, filename):
        records = []

        for name in self.shape_names:
            records.append({
                "name": name,
                "shape": self._shape_to_dict(self.shapes[name])
            })

        with open(filename, "w") as f:
            json.dump(records, f, indent=2)

    # Load a drawing from its saved JSON file
    @classmethod
    def load(cls, filename):
        with open(filename, "r") as f:
            records = json.load(f)

        drawing = cls()

        for item in records:
            shape = cls._dict_to_shape(item["shape"])
            drawing.add_shape(shape, item["name"])

        return drawing
```
