### Command-line basics

There is plenty of material online that will help you figure out how to do various tasks on the command line. Commands you may need to know today:

* `ls`: lists the contents of the current directory.
* `pwd`: prints the path of the current directory.
* `cd <directory>`: changes your current directory to the specified directory.
* `cd ..`: changes current directory to the previous directory. Basically steps out of the current directory to the directory containing the current directory.
* `mkdir <directory>`: create a new directory with the specified name.
* `rmdir <directory>`: removes the specified directory. Note it has to be empty.
* `rm <filename>`: deletes the specified file.
* `mv <filename 1> <filename 2>`: Moves or renames a file.
* `cp <filename 1> <filename 2>`: copies an file. If you just provide a path to a directory, it copies the file into that directory with the same filename. If you specifiy a new filename, the copy has a new name. For example `cp File.1.txt File.2.txt` creates a copy of `File.1.txt` with the name `File.2.txt`. Meanwhile `cp File.1.txt my_directory`, where `my_directory` is a directory, creates a copy of `File.1.txt` in directory `my_directory` with the name `File.1.txt`.

````python
# Skeleton

# Created a function that generates a list of random numbers between a minimum and maximum value.






def generate_uniform(N,x_min,x_max):
    out = []
    ### BEGIN SOLUTION

    #generate random numbers
    for i in range(N):
        # Scale each random number to the specified range
        value = x_min + random.random() * (x_max - x_min)

        # Add the value to our list
        out.append(value)

    
    ### END SOLUTION
    return out
````

Write a function that computes the mean of values in a list. Recall the equation for the mean of a random variable $\bf{x}$ computed on a data set of $n$ values $\{ x_i \} = \{x_1, x_2, ..., x_n\}$  is ${\bf\bar{x}} = \frac{1}{n} \sum_i^n x_i$.


I wrote a function which allows me to calculate the means of a list of numbers. Using a for loop to add all values together, then dividing the total by number of values in the list.

````python
# Skeleton
def mean(Data):
    m=0.
    
    ### BEGIN SOLUTION

    # Add all the numbers together
    for number in Data:
        m += number

    # Divide by how many numbers are in the list
    m = m / len(Data)       
    
    ### END SOLUTION
    
    return m
````

*Exercise 2b:* 
Write a function that computes the variance of values in a list. Recall the equation for the variance of a random variable $\bf{x}$ computed on a data set of $n$ values $\{ x_i \} = \{x_1, x_2, ..., x_n\}$  is $s^2 = \frac{1}{n-1} \sum_i^n (x_i - {\bf\bar{x}})^2$.

(Note the notation: $\langle x \rangle$ means the *mean*, so the variance is written $s^2$ — or
$\sigma^2$ when it is the variance of a distribution rather than of a sample. The $n-1$ rather
than $n$ is the Bessel correction: you had to estimate $\bf\bar{x}$ from the same data, which
spends one degree of freedom.)

````python
# Skeleton

# created a function to calculate the sample variance
# Calculated mean of data and then using a loop found and squared the difference between each value and mean
# then divided the toal by n-1 to get sample variance.

def variance(Data):
    m=0.
    
    ### BEGIN SOLUTION

    average = mean(Data)

    # Add up the squared differences from the mean
    for number in Data:
        difference = number - average
        m += difference ** 2

    # Divide by n - 1 for sample variance
    m = m / (len(Data) - 1)       
    
    ### END SOLUTION
    
    return m
````

````python
# Test your solution here
print ("Variance of Data:", variance(data))
````

## Histogramming

*Exercise 3:* Write a function that bins the data so that you can create a histogram. An example of how to implement histogramming is the following logic:

* User inputs a list of values `x` and optionally `n_bins` which defaults to 10.
* If not supplied, find the minimum and maximum (`x_min`,`x_max`) of the values in x.
* Determine the bin size (`bin_size`) by dividing the range of the function by the number of bins.
* Create an empty list of zeros of size `n_bins`, call it `hist`.
* Loop over the values in `x`
    * Loop over the values in `hist` with index `i`:
        * If x is between `x_min+i*bin_size` and `x_min+(i+1)*bin_size`, increment `hist[i].` 
        * Once you have found the right bin there is no reason to keep checking the rest, so use `break` to stop the inner loop and move on to the next data point.
* Return `hist` and the list corresponding of the bin edges (i.e. of `x_min+i*bin_size`).


- I created a function which divided the dataset into bins to make a histogram. I first calculated the minimum and maximum values and then used those to find the width of each bin. I used a nested loop to check the data value against the bin boundaries and increase the corresponding count. The function returns a result of bin counts and edges and you can use that to display the distribution of the data.

````python
# Solution
def histogram(x,n_bins=10,x_min=None,x_max=None):
    ### BEGIN SOLUTION

        # Find the minimum and maximum
    if x_min is None:
        x_min = min(x)

    if x_max is None:
        x_max = max(x)

    # Calculate the width of each bin
    bin_size = (x_max - x_min) / n_bins

    # Create a list of zeros to count the values
    hist = [0] * n_bins

    # Store the edges of the bins
    bin_edges = []

    for i in range(n_bins + 1):
        edge = x_min + i * bin_size
        bin_edges.append(edge)

    # Check which bin each number belongs to
    for number in x:
        for i in range(n_bins):

            lower = bin_edges[i]
            upper = bin_edges[i + 1]

            # Include the max value 
            if lower <= number < upper or (i == n_bins - 1 and number == upper):
                hist[i] += 1
                break

    
    ### END SOLUTION

    return hist,bin_edges
````


Where each line corresponds to a bin and the number of `#`'s are proportional to the value of the data in the bin.

That example is only there to show the *format* — the bin ranges shown don't match our `data`,
which runs from -10 to 10. When you run it on `data` you should get ten or twenty bars of roughly
equal length, because `data` is uniform. **That flatness is the check**: if your bars are not
roughly equal, either the generator or the histogram is wrong. 

````python
# Solution

# Used the histogram finction from before to create a text based histogram and calculate number of values in each bin
# Used string multiplication to create the bars, and scaled them to largest bin count so histogram would fit within ouput.




def draw_histogram(x,n_bins,x_min=None,x_max=None,character="#",max_character_per_line=20):
    ### BEGIN SOLUTION

    hist, bin_edges = histogram(x, n_bins, x_min, x_max)

    # Find the largest count so we can scale the bars
    max_count = max(hist)

    # Go through each bin and print a bar
    for i in range(n_bins):

        # Get the lower and upper limits of the bin
        lower = bin_edges[i]
        upper = bin_edges[i + 1]

        # Calculate how many characters to print
        if max_count > 0:
            bar_length = round(
                hist[i] / max_count * max_character_per_line
            )
        else:
            bar_length = 0

        # Create the bar using the chosen character
        bar = character * bar_length

        # Print the bin range and its bar
        print(f"[{lower:6.2f}, {upper:6.2f}] : {bar}")
    
    ### END SOLUTION

    return hist,bin_edges
````

*Exercise 5:* Write a function that applies a boolean function (one that returns true/false) to every element in data, and return a list of indices of elements where the result was true. Use this function to find the indices of entries greater than 0.5. 

````python
# Created a function which applies a condition to every value in a list.
# I checked each value with a loop and stored the index when return is true.
# tested this by finding values greater than 0.5




def where(mylist,myfunc):
    out= []
    
    ### BEGIN SOLUTION

    # Go through each position in the list
    for i in range(len(mylist)):

        # Check if the value meets the condition
        if myfunc(mylist[i]) == True:
            out.append(i)

            
    ### END SOLUTION
    
    return out
````

*Exercise 8:* Write a "generator" function called `generate_function(func,x_min,x_max,N)`, that instead of generating a flat distribution, generates a distribution with functional form coded in `func`. Note that `func` will always be > 0.  

Use the test function below and your histogramming functions above to demonstrate that your generator is working properly.

Hint: A simple, but slow, solution is to a draw random number `test_x` within the specified range and another number `p` between the `min` and `max` of the function (which you will have to determine). If `p<=function(test_x)`, then place `test_x` on the output. If not, repeat the process, drawing two new numbers. Repeat until you have the specified number of generated numbers, `N`. For this problem, it's OK to determine the `min` and `max` by numerically sampling the function.  



-In this exercise I create a function that generates random numbers based on the shape of another function. I used a for loop to estimate the max hwight of the function and while loop to generate numbers until the requested amount. The absolute value function produced the correct V-shaped graph.

````python
def generate_function(func,x_min,x_max,N=1000):
    out = list()
    ### BEGIN SOLUTION

    # Find an approximate maximum of the function
    max_y = 0

    for i in range(1001):
        x = x_min + i * (x_max - x_min) / 1000
        y = func(x)

        if y > max_y:
            max_y = y

    # Generate numbers until the list contains N values
    while len(out) < N:

        # Pick a random x value in the range
        test_x = x_min + random.random() * (x_max - x_min)

        # Pick a random height between 0 and the maximum
        p = random.random() * max_y

        # Keep the x value if it is below the function
        if p <= func(test_x):
            out.append(test_x)       
    
    ### END SOLUTION
    
    return out
````

*Exercise 10:* Combine your `generate_function`, `where`, and `in_range` functions above to
create an integrate function.

To be precise about what it should return: **the fraction of the distribution that lies between
`x_min` and `x_max`.** Because `gaussian` is normalised (it integrates to 1 over all $x$), that
fraction *is* the integral of the distribution over that interval.

The way to get it: use `generate_function` to draw `n_points` numbers from `func`, then use
`where` and `in_range` to count how many landed between `x_min` and `x_max`. The answer is that
count divided by `n_points`.

One wrinkle — `generate_function` needs a range to generate over, and it has to be wide enough to
contain essentially the whole distribution, not just the interval you are integrating. For a
Gaussian, mean $\pm\,5\sigma$ is plenty. Feel free to add arguments to the function for that.

Use it to show that approximately 68% of a Normal distribution lies within one standard deviation
of the mean.

````python
# Skeleton
def integrate(func, x_min, x_max, n_points=1000):
    ### BEGIN SOLUTION

    # Generate random numbers across a wider range
    # For our standard normal test, -5 to 5 covers
    # almost the entire distribution.
    data = generate_function(func, -5, 5, n_points)

    # Find the positions of numbers inside the interval
    positions = where(data, in_range(x_min, x_max))

    # Count the numbers inside the interval
    count = len(positions)

    # Divide by the total number of random points
    integral = count / n_points

    
    ### END SOLUTION
    return integral
````

````python
# Test your solution here
g=gaussian(0,1)

print ("Within one sigma:", integrate(g,-1,1), " (should be close to 0.68)")
print ("Within two sigma:", integrate(g,-2,2), " (should be close to 0.95)")
````

````python
# Skeleton
def estimate_spread(N,n_trials=20):
    estimates = []
    ### BEGIN SOLUTION

        # Create a standard normal distribution
    g = gaussian(0, 1)

    # Repeat the integration several times
    for i in range(n_trials):
        result = integrate(g, -1, 1, N)

        # Save each estimate in a list
        estimates.append(result)
    
    ### END SOLUTION
    return mean(estimates), variance(estimates)
````

*Exercise 12:* Your `generate_function` works on any function you can evaluate, but it throws
draws away. When you *can* do the integral by hand, there is a method with no waste at all:
**inversion**.

The exponential distribution is

$$P(t) = \frac{1}{\tau}e^{-t/\tau} \quad (t \ge 0)$$

Its cumulative is $F(T) = 1 - e^{-T/\tau}$. Setting that equal to a uniform random number $u$ and
solving for $T$ gives

$$T = -\tau \ln(1-u)$$

**(a)** Write `generate_exp(tau,N)` that returns `N` exponentially distributed numbers using that
one line. It needs `random.random()` and `math.log` — no loop over rejected draws.

**(b)** Check it: the mean *and* the standard deviation of an exponential are both $\tau$. Use
your `mean` and `variance` functions.

**(c)** Now compare the cost. Use `generate_function` to draw the same distribution by
accept/reject over the range 0 to $5\tau$, counting how many random draws it needs per number it
keeps. Inversion needs one. How many does accept/reject need, and why that many?

````python
# Skeleton
import math

def generate_exp(tau,N):
    out = []
    ### BEGIN SOLUTION

    

    for i in range(N):
        # Generate a random number between 0 and 1
        u = random.random()

        # Use the inverse formula for the exponential distribution
        t = -tau * math.log(1 - u)

        # Store the value in our list
        out.append(t)
        
    
    ### END SOLUTION
    return out
````

````python


# Exercise 12(c)
# Compare inverse transform and acceptance-rejection methods.

import random
import math

# Define the number of values and distribution parameter
N = 10000
tau = 2.0

def exponential(x):
    return (1 / tau) * math.exp(-x / tau)

# Maximum height of the exponential distribution
max_y = 1 / tau

accepted = []
attempts = 0

# Generate values using acceptance-rejection
while len(accepted) < N:
    attempts += 1

    # Generate a random x value
    x = random.random() * (5 * tau)

    # Generate a random height
    y = random.random() * max_y

    # Accept the value if it falls below the curve
    if y <= exponential(x):
        accepted.append(x)

# Calculate efficiency
attempts_per_value = attempts / N
draws_per_value = 2 * attempts / N

print("Inverse method: 1 random draw per value")
print("Acceptance-rejection attempts:", attempts)
print("Attempts per accepted value:", attempts_per_value)
print("Random draws per accepted value:", draws_per_value)
````

````python
# Skeleton
def percentile(x,fraction,n_bins=100):
    ### BEGIN SOLUTION

    counts, edges = histogram(x, n_bins)

    # Find how many values we need to reach
    target = fraction * len(x)

    # Keep track of the total number of values
    total = 0

    # Start with the first bin edge
    x_p = edges[0]

    # Go through the bins and add the counts
    for i in range(n_bins):
        total += counts[i]

        # Stop when we reach the requested percentage
        if total >= target:
            x_p = edges[i + 1]
            break

    
    ### END SOLUTION
    return x_p
````
